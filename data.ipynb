## EDA

```python
plt.Figure(figsize=(16,8))
sns.countplot(data=df,x='income')
plt.title("Income class balance")
plt.xlabel("Income")
plt.ylabel("Number of people")
plt.show()
```

```python
plt.Figure(figsize=(16,8))
sns.countplot(data=df,x='sex')
plt.title("Number of men and women")
plt.xlabel("Sex")
plt.ylabel("Number of people")
plt.show()
```

```python
plt.Figure(figsize=(16,8))
sns.barplot(data=df,x='sex',y='income')
plt.title('Income Distribution on Gender')
plt.xlabel('Number of men and women')
plt.ylabel('Income Distribution')
plt.show()
```

```python
plt.Figure(figsize=(16,8))
sns.countplot(data=df,x='workclass')
plt.title("Number of people in each workclass")
plt.xlabel("Workclass")
plt.ylabel("Number of people")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.Figure(figsize=(16,8))
sns.barplot(data=df,x='workclass',y='income')
plt.title('Income Distribution on different classes')
plt.xlabel('Number of classes')
plt.ylabel('Income Distribution')
plt.xticks(rotation=45)
plt.show()
```

```python
top=df['native_country'].value_counts().head(10).index
top
```

```python
plt.Figure(figsize=(8,6))
sns.barplot(data=df[df['native_country'].isin(top)],x='native_country',y='income')
plt.title('Share earning >50K by country (top 10 by count)')  # chart title
plt.xlabel('Share earning >50K')                      # x-axis label (the number is on x here)
plt.xticks(rotation=45)
plt.ylabel('Native country')                          # y-axis label (the names are on y here)
plt.show()  
```

```python
plt.figure(figsize=(8,6))
sns.barplot(data=df,x='income',y='age')
plt.title("Age by income class")
plt.xlabel('Income (0 = <=50K, 1 = >50K)')            # x-axis label
plt.ylabel('Age')                                     # y-axis label
plt.show()     
```

```python
plt.figure(figsize=(8,6))
sns.barplot(data=df,x='sex',y='hours_per_week')
plt.title(" WORKING HOURS OF BOTH GENDER")
plt.xlabel('GENDER')            
plt.ylabel('AGE')                                     
plt.show()     
```

```python
plt.figure(figsize=(7, 5))                            # chart size
sns.boxplot(data=df, x='sex', y='hours_per_week')     # working hours spread for each sex
plt.title('Working hours per week by sex')            # chart title
plt.xlabel('Sex')                                     # x-axis label
plt.ylabel('Hours per week')                          # y-axis label
plt.show()                                            # display the chart
```

```python
sns.heatmap(df.corr(numeric_only=True),fmt='.2g',annot=True,cmap='coolwarm')
plt.title('Correlation between number columns')       # chart title
plt.xlabel('Columns')                                 # x-axis label
plt.ylabel('Columns')                                 # y-axis label
plt.show()   
```
